# Data Inspection

This notebook reviews the structure and basic contents of the raw Statistics Canada datasets.

The inspection focuses on:

- File size
- CSV structure
- Column names
- Sample rows
- Row structure consistency

## Purpose

The raw datasets are large CSV files, so they are inspected without loading the full files into memory.

The detailed inspection is performed by `02_inspect.py`.

## Key Observations

Record only findings that are relevant to later data preparation, such as:

- Dataset file sizes
- Number of columns
- Column naming patterns
- Presence of consistent CSV structure
- Any issues requiring attention during extraction or transformation
