# Transformation Design

## 1. Purpose

This project examines labour market tightness in Canada by connecting employment, wages, and job vacancy indicators.

The analysis aims to identify whether changes in labour demand are accompanied by changes in employment and wages across industries.

## 2. Target grain

The target analytical grain is:

* Canada × Industry × Month

* The primary analysis period is January 2021 to December 2025.

* The analysis will use Canada-level data and existing Statistics Canada industry classifications.

## 3. Source datasets

The analysis uses three monthly Statistics Canada datasets:

* Labour force characteristics by industry: employment and labour force indicators

* Employee wages by industry: wage indicators

* Job vacancies, payroll employees and job vacancy rates by industry: labour demand indicators

The datasets provide related labour market measures but differ in structure and available dimensions.

## 4. Dimension and data alignment

The datasets will be independently aligned to the target analytical grain using:

* **Canada** as the common geography

* **NAICS** as the industry classification

* **Month** as the time dimension

* Selected measures relevant to each dataset

The three datasets will remain as separate analytical tables after transformation. They will not be directly joined during the transformation step.

The common dimensions will allow the datasets to be compared or joined later during analysis when appropriate.

Raw NAICS labels will not be joined directly until the industry dimension has been standardized.

Transformations will preserve the original Statistics Canada classifications wherever possible.

## 5. Data quality rules

The analysis will apply the following rules:

* Treat `x` as suppressed and exclude it from numerical analysis.

* Treat `F` as too unreliable to publish and exclude it from analysis.

* Filter to Canada and the primary analysis period of **2021-01 to 2025-12**.

## 6. Measures to retain

Only measures directly relevant to the analysis will be retained:

* Employment: labour market outcome

* Wages: compensation outcome

* Job vacancies: labour demand

* Job vacancy rate: labour market tightness

Additional dimensions will only be retained when required to identify or interpret these measures.

## 7. Output table design

The transformation process will produce three separate analytical tables:

* **Employment table**: Canada × Industry × Month

* **Wages table**: Canada × Industry × Month

* **Vacancies table**: Canada × Industry × Month

Each row in each table will represent one industry in one month, with the selected measures for that dataset as columns.

The three tables will share common analytical dimensions but will remain separate.

The processed tables will be stored in `data/processed/` for downstream analysis and visualization in SQL and Power BI.

## 8. Transformation requirements

The transformation process will:

* Standardize column names and data types.

* Standardize the industry column as `NAICS`.

* Normalize equivalent NAICS label formats before analysis:

  * `[55, 56]` → `[55-56]`
  * `[52, 53]` → `[52-53]`

* Preserve the original Statistics Canada classifications and values wherever possible.

* Filter each dataset to Canada and the primary analysis period.

* Apply the defined data quality rules.

* Select and reshape the required measures for each dataset.

* Align each dataset independently to the target grain.

* Keep the three transformed datasets separate.

* Validate that each resulting table matches the target grain.

* Save the transformed datasets to `data/processed/`.
