# Metadata Inspection

## Objective

Inspect the Statistics Canada metadata files to understand the structure, dimensions, reference periods, and key data-quality notes of the three datasets.

## Dataset Overview

| Dataset  | Main Content                                       | Dimensions | Reference Period   |
| -------- | -------------------------------------------------- | ---------: | ------------------ |
| 14100022 | Labour force characteristics by industry           |          5 | 1976-01 to 2026-08 |
| 14100063 | Employee wages by industry                         |          6 | 1997-01 to 2026-08 |
| 14100372 | Job vacancies, payroll employees, and vacancy rate |          3 | 2015-04 to 2026-06 |

All three datasets are monthly and unadjusted for seasonality.

## Key Findings

* All three datasets include **NAICS** as an industry dimension.
* The datasets have different dimensional structures and therefore different grains.
* 14100022 and 14100063 include demographic dimensions such as gender and age group.
* 14100372 contains job vacancies, payroll employees, and vacancy rate, and is available at the **Canada level** in the extracted data.
* The datasets have different reference-period coverage.
* Metadata identifies suppressed and unavailable values that must be handled explicitly during data preparation.
* 14100372 monthly estimates may be revised when quarterly estimates are released.

## Implication

NAICS appears to be the main potential link across the three datasets, but **NAICS levels and dataset grain must be compared before joining**.

The primary analysis period will be **2021-01 to 2025-12**, with 2026 data treated as reference data where applicable.
