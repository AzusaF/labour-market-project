# EDA
## 1. EDA Objective
The objective of this EDA is to assess the structure, analytical grain, compatibility, and data quality of the three Statistics Canada datasets before transformation and integration.

The EDA focuses on identifying:
- the analytical grain and dimensions of each dataset;
- inconsistencies in NAICS industry classifications;
- compatibility of geography and time coverage;
- missing values, suppression, and data-quality indicators; and
- the requirements for creating a consistent analytical dataset for subsequent analysis.

## 2. Key Analytical Decisions

| Item                        | Decision                                                                                                                     |
| --------------------------- | ---------------------------------------------------------------------------------------------------------------------------- |
| **Target grain**            | Canada × Industry × Month                                                                                                    |
| **Geography**               | Canada                                                                                                                       |
| **Primary analysis period** | January 2021 – December 2025                                                                                                 |
| **NAICS handling**          | Raw industry labels will not be used for direct joins; industries will be standardized through an explicit mapping strategy. |
| **Suppression handling**    | Suppressed values (`x`) will be treated as unavailable, not as zero.                                                         |

## 3. Dataset Overview
| Dataset  | Main measure                                     | Grain                                                   | Geography          |
| -------- | ------------------------------------------------ | ------------------------------------------------------- | ------------------ |
| 14100022 | Employment / labour force characteristics        | Month × Geography × Industry × Labour-force dimension   | Canada + provinces |
| 14100063 | Employee wages                                   | Month × Geography × Industry × Work type × Wage measure | Canada + provinces |
| 14100372 | Job vacancies / payroll employees / vacancy rate | Month × Industry                                        | Canada only        |

### Finding
The three datasets are related conceptually, but they do not share the same analytical grain.

## 4. NAICS Structure

### Finding
The datasets do not provide a fully consistent NAICS structure. Although all three datasets contain industry information, the available industry categories and levels differ across datasets. Direct joins using industry labels are therefore not considered reliable without an explicit industry mapping strategy.

### Decision
- Use an explicit NAICS mapping strategy before cross-dataset joins.  

Industry categories will be standardized to a common analytical level before integration. The mapping strategy and target industry level will be documented in the transformation stage.
## 5. Grain Compatibility

### Finding
14100022: Geography × Industry × Labour Force Measure × Month  
14100063: Geography × Industry × Work Type × Wage Measure × Month  
14100372: Canada × Industry × Month  

A direct three-way join is not appropriate because the datasets contain different dimensions beyond the shared time and industry dimensions. Dataset-specific dimensions must be aggregated or filtered before joining.

### Decision
- Use Canada × Industry × Month as the target analytical grain. 

Dataset-specific dimensions will be filtered or aggregated before integration so that each dataset contributes one or more metrics at the target grain.

## 6. Geography Compatibility

### Finding
Job vacancy data (14100372) is available at the Canada level, while the other datasets provide both Canada and provincial observations.

### Decision
- Use Canada-level observations for the primary cross-dataset analysis. 

Provincial observations will not be included in the integrated dataset because 14100372 does not provide comparable provincial job vacancy data.

## 7. Time Coverage

### Finding
The datasets do not have identical time coverage. To maintain a consistent comparison period, the primary analysis will use the common period from January 2021 through December 2025.

### Decision
- Use January 2021 through December 2025 as the primary analysis period. 

Data from 2026 will be retained as a reference where available but excluded from the primary cross-dataset comparison.

## 8. Data Quality and Suppression

### Finding
Missing values are not equivalent to random missing observations. In particular, `x` represents values suppressed for confidentiality and should not be imputed as zero.  
The job vacancy dataset includes a quality/status indicator. Observations marked `F` indicate that the estimate is too unreliable to publish.

### Decision
- Preserve suppression and quality information during transformation. 

Suppressed values (`x`) will be treated as unavailable rather than zero and will not be imputed. Observations marked F in 14100372 will be excluded from quantitative analysis where applicable, consistent with Statistics Canada’s data-quality guidance. Data-quality handling will be documented in the analysis workflow.

## 9. Final Analytical Design

### Target Grain
Canada × Industry × Month

### Join Strategy
Each dataset will first be transformed independently to the target grain. Dataset-specific dimensions such as labour-force measures, work types, and wage measures will be filtered or aggregated as appropriate. The resulting datasets will then be joined using Month and the standardized NAICS industry mapping.

### Analysis Period
January 2021 through December 2025

### Key Metrics
- Employment / labour force measure from 14100022
- Employee wage measure from 14100063
- Job vacancies from 14100372
- Payroll employees from 14100372
- Job vacancy rate from 14100372

The analysis will focus on changes and relationships among employment, wages, job vacancies, payroll employment, and vacancy rates across industries over time.

## 10. EDA Summary
The EDA established that the three Statistics Canada datasets are conceptually related but require structural harmonization before integration. The main challenges are inconsistent NAICS structures, different analytical grains, different geographic coverage, and dataset-specific data-quality considerations.

Based on these findings, the analysis will use Canada × Industry × Month as the common analytical grain and January 2021 through December 2025 as the primary comparison period. Each dataset will be transformed independently before being joined through a documented NAICS mapping strategy. Suppressed and unreliable observations will be preserved and handled explicitly rather than treated as zero or valid estimates.

This design provides a consistent foundation for the subsequent transformation, database integration, and analysis stages.
